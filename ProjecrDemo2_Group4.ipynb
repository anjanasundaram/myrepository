```python
import time
# Product Class
class Product:
    def __init__(self, product_id, name, quantity, category):
        self.product_id = product_id
        self.name = name
        self.quantity = quantity
        self.category = category

    def __str__(self):
        return f"ID: {self.product_id}, Name: {self.name}, Quantity: {self.quantity}, Category: {self.category}"


# Dynamic Array Class for Product Inventory
class DynamicArray:
    def __init__(self):
        self.array = [None]
        self.size = 0
        self.capacity = 1

    def resize(self, new_capacity):
        new_array = [None] * new_capacity
        for i in range(self.size):
            new_array[i] = self.array[i]
        self.array = new_array
        self.capacity = new_capacity

    def add_product(self, product):
        if self.size == self.capacity:
            self.resize(self.capacity * 2)
        self.array[self.size] = product
        self.size += 1

    def delete_product(self, product_id):
        index_to_delete = -1
        for i in range(self.size):
            if self.array[i].product_id == product_id:
                index_to_delete = i
                break

        if index_to_delete == -1:
            print("Product does not exist")
            return

        for i in range(index_to_delete, self.size - 1):
            self.array[i] = self.array[i + 1]

        self.array[self.size - 1] = None
        self.size -= 1

        if self.size <= self.capacity // 4:
            self.resize(max(1, self.capacity // 2))

    def update_product(self, product_id, new_name=None, new_quantity=None, new_category=None):
        for i in range(self.size):
            if self.array[i].product_id == product_id:
                if new_name:
                    self.array[i].name = new_name
                if new_quantity is not None:
                    self.array[i].quantity = new_quantity
                if new_category:
                    self.array[i].category = new_category
                return
        print("There is no product to update")

    def display_product(self):
        if self.size == 0:
            print("The Warehouse is Empty")
        else:
            for i in range(self.size):
                print(self.array[i])

    def merge_sort_by(self, key):
        def merge(left, right):
            sorted_list = [] #stores the list after has merged it
            i = j = 0 #starts at index 0 for both left and right
            while i < len(left) and j < len(right):
                if getattr(left[i], key) <= getattr(right[j], key): # access the values of the attributes specified by the key
                    sorted_list.append(left[i])# adds the smaller item to the sorted list
                    i += 1 # moves to the next item on the left
                else:
                    sorted_list.append(right[j])# adds the smaller item from the right
                    j += 1  # moves to the next item on the left
            sorted_list.extend(left[i:]) # adds the rest of the items on the left
            sorted_list.extend(right[j:]) #adds the rest of the items on the right
            return sorted_list # returns the final merged and sorted list after divide and conquer

        def merge_sort(arr): # this is how we recursively sort and merge the elements
            if len(arr) <= 1: #Check the size of array and return if it is less than or equal to 1
                return arr
            mid = len(arr) // 2  # splits the list into 2
            left = merge_sort(arr[:mid])   # Recursively sort the left half of the array
            right = merge_sort(arr[mid:])  # Recursively sort the right half of the array
            return merge(left, right)

        valid_items = self.array[:self.size] # Ignores empty spaces beyond `self.size`
        sorted_array = merge_sort(valid_items) # Perform merge sort on the list of valid (non-null) items and store the sorted result
        for i in range(self.size):
            self.array[i] = sorted_array[i] #putting back all the items in the original array
```

```python
from collections import defaultdict, deque

# Warehouse Graph implementation
class WarehouseGraph:
    def __init__(self):
        self.graph = defaultdict(list)
        self.section_products = defaultdict(DynamicArray)

    def add_section(self, section_name):
        if section_name not in self.graph:
            self.graph[section_name] = []

    def add_path(self, section1, section2):
        self.add_section(section1)
        self.add_section(section2)
        self.graph[section1].append(section2)
        self.graph[section2].append(section1)

    def add_product_to_section(self, section, product):
        self.add_section(section)
        self.section_products[section].add_product(product)

    def get_products_in_section(self, section):
        return self.section_products.get(section, DynamicArray())

    def print_warehouse_graph_text(self, root="Warehouse", indent="", visited=None):
      """
      Recursively prints a text-based graph representation:
      Warehouse (root) → Sections → Products (including nested child sections).
      """
      if visited is None:
          visited = set()
          print("\n" + root)

      visited.add(root)
      neighbors = self.graph.get(root, [])

      for i, section in enumerate(sorted(neighbors)):
          if section in visited:
              continue  # Prevent revisiting in case of cycles

          is_last_section = i == len(neighbors) - 1
          section_prefix = indent + ("└── " if is_last_section else "├── ")
          print(section_prefix + section)

          product_indent = indent + ("    " if is_last_section else "│   ")

          # Print products in this section
          products = self.get_products_in_section(section)
          for j in range(products.size):
              product = products.array[j]
              is_last_product = j == products.size - 1
              prod_prefix = product_indent + ("└── " if is_last_product else "├── ")
              print(prod_prefix + f"[{product.category}]")

          # Recursively print children of this section
          self.print_warehouse_graph_text(section, indent=product_indent, visited=visited)


    def print_section_connections(self, root="Warehouse"):
      """
      Prints the warehouse section-to-section connections, highlighting the warehouse as the root node.
      Shows each section and the sections it connects to, skipping the root from child nodes.
      """
      print(f"\n📦 Warehouse Layout (Root: {root})")
      for section, neighbors in self.graph.items():
          # Skip showing root as child in any section
          filtered_neighbors = [n for n in neighbors if not (section != root and n == root)]
          connections = ', '.join(filtered_neighbors)

          if section == root:
              print(f"{section} → {connections}")
          else:
              print(f"{section} → {connections}")


    def bfs_shortest_path_to_product(self, start_section, product_id):
      visited = set()  # Keeps track of visited sections to avoid cycles
      queue = deque()  # Queue for BFS traversal
      queue.append((start_section, [start_section]))  # Start from the given section with initial path

      while queue:
          current_section, path = queue.popleft()  # Get the next section and the path to reach it

          # Check if the current section contains the product
          products = self.get_products_in_section(current_section)
          for i in range(products.size):
              if products.array[i] and products.array[i].product_id == product_id:
                  return path  # Product found — return the path taken to reach it

          # Add unvisited neighboring sections to the queue
          for neighbor in self.graph[current_section]:
              if neighbor not in visited:
                  visited.add(neighbor)
                  new_path = path.copy()  # Create a new path by extending the current one
                  new_path.append(neighbor)
                  queue.append((neighbor, new_path))  # Enqueue the neighbor with its path

      return None  # Return None if product not found in any reachable section

    def dfs_products_in_category(self, start_section, category):
      visited = set()  # Track visited sections to prevent revisiting
      products_found = DynamicArray()  # Stores all matching products found

      def dfs_helper(section):
          visited.add(section)  # Mark the current section as visited

          # Check for products in the current section that match the category
          products = self.get_products_in_section(section)
          for i in range(products.size):
              if products.array[i] and products.array[i].category == category:
                  products_found.add_product(products.array[i])  # Add matching product to result

          # Recursively explore connected sections
          for neighbor in self.graph[section]:
              if neighbor not in visited:
                  dfs_helper(neighbor)

      dfs_helper(start_section)  # Start DFS from the initial section
      return products_found  # Return the list of all matching products


# Add products
warehouse = WarehouseGraph()

# Add sections and paths
# warehouse.add_path("Loading Dock", "Section A")
# warehouse.add_path("Section A", "Section B")
# warehouse.add_path("Section A", "Section C")
# warehouse.add_path("Section B", "Section D")
# warehouse.add_path("Section C", "Section E")

# # PackageType section
warehouse.add_product_to_section("PackageType", Product("PKG001", "Parcel to Toronto", 1, "Parcel"))
warehouse.add_product_to_section("PackageType", Product("PKG022", "Oversize Parcel to Calgary", 2, "Oversize"))
warehouse.add_product_to_section("PackageType", Product("PKG023", "Fragile Glassware to Montreal", 1, "Fragile"))
warehouse.add_product_to_section("PackageType", Product("PKG024", "Cold Storage - Vaccine Shipment", 3, "ColdStorage"))
warehouse.add_product_to_section("PackageType", Product("PKG025", "Hazardous Chemicals to Edmonton", 2, "Hazardous"))

#Parcel type section
warehouse.add_product_to_section("Parcel", Product("PKG201", "Letter to Toronto", 1, "Letter"))
warehouse.add_product_to_section("Parcel", Product("PKG202", "Package that weighs more than 1 kg", 1, "Package"))
warehouse.add_product_to_section("Letter", Product("PKG211", "Letter to Toronto", 1, "Flat"))
warehouse.add_product_to_section("Letter", Product("PKG212", "Letter to Toronto", 1, "Envelope"))

# Medical
warehouse.add_product_to_section("Medical", Product("PKG002", "Hazardous Chemicals to Edmonton", 5, "Hazardous"))
warehouse.add_product_to_section("Medical", Product("PKG026", "Medicine requires cold storage", 4, "ColdStorage"))
warehouse.add_product_to_section("Medical", Product("PKG027", "Fragile Medical equipment parts", 5, "Fragile"))

# Electronics
warehouse.add_product_to_section("Electronics", Product("PKG101", "Hazardous Chemicals to Edmonton", 5, "Hazardous"))
warehouse.add_product_to_section("Electronics", Product("PKG102", "Fragile Medical equipment parts", 5, "Fragile"))
warehouse.add_product_to_section("Electronics", Product("PKG103", "Oversize Parcel to Calgary", 2, "Oversize"))

# shipment handling section
warehouse.add_product_to_section("ShipmentHandling", Product("PKG004", "Express Document", 3, "Return"))
warehouse.add_product_to_section("ShipmentHandling", Product("PKG033", "Letter to Medicine Hat", 2, "Deliver"))
warehouse.add_product_to_section("ShipmentHandling", Product("PKG034", "Package to Brampton", 3, "Pickup"))
warehouse.add_product_to_section("ShipmentHandling", Product("PKG035", "Parcel to Windsor", 5, "on-hold"))
warehouse.add_product_to_section("ShipmentHandling", Product("PKG036", "Parcel to Windsor", 2, "Telegrams"))

# Shipment speed
warehouse.add_product_to_section("ShipmentSpeed", Product("PKG005", "Large Parcel", 2, "Priority"))
warehouse.add_product_to_section("ShipmentSpeed", Product("PKG039", "Express Document to Kingston", 2, "Xpresspost"))
warehouse.add_product_to_section("ShipmentSpeed", Product("PKG037", "Oversize Freight to Sudbury", 1, "Flat rate box"))
warehouse.add_product_to_section("ShipmentSpeed", Product("PKG038", "Medical Equipment to Saskatoon", 3, "Regular Parcel"))

warehouse.add_path("Warehouse", "PackageType")
warehouse.add_path("PackageType", "Medical")
warehouse.add_path("PackageType", "Parcel")
warehouse.add_path("PackageType", "Electronics")
warehouse.add_path("Parcel", "Letter")
warehouse.add_path("Warehouse", "ShipmentHandling")
warehouse.add_path("Warehouse", "ShipmentSpeed")

warehouse.print_section_connections()

warehouse.print_warehouse_graph_text()
```

```python
class TreeNode:
    def __init__(self, value):
        self.value = value
        self.children = []

    def add_child(self, child_node):
        self.children.append(child_node)

    def __repr__(self):
        return str(self.value)

def print_tree(node, level=0, prefix="Root: "):
    """ASCII visualization of the tree"""
    print("    " * level + prefix + str(node.value))
    for i, child in enumerate(node.children):
        is_last = i == len(node.children) - 1
        new_prefix = "└── " if is_last else "├── "
        print_tree(child, level + 1, new_prefix)

# Example usage
root = TreeNode("Warehouse")
section_a = TreeNode("Section A")
section_b = TreeNode("Section B")
section_c = TreeNode("Section C")

root.add_child(section_a)
root.add_child(section_b)
section_a.add_child(section_c)

print("ASCII Tree Visualization:")
print_tree(root)
```
